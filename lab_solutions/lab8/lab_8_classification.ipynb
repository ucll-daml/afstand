#### Our next case: Hotel booking dataset

source: https://www.sciencedirect.com/science/article/pii/S2352340918315191

*This data article describes two datasets with hotel demand data. One of the hotels (H1) is a resort hotel and the other is a city hotel (H2). Both datasets share the same structure, with 31 variables describing the 40,060 observations of H1 and 79,330 observations of H2. Each observation represents a hotel booking. Both datasets comprehend bookings due to arrive between the 1st of July of 2015 and the 31st of August 2017, including bookings that effectively arrived and bookings that were canceled. Both hotels are located in Portugal: H1 at the resort region of Algarve and H2 at the city of Lisbon.*

The goal is to help the two hotels maximize their revenue.

## Data Preparation

#### ❓Go back to lab 6 and take another look on how to prepare your data for modeling. The same steps are needed for classification as for regression.

```python
from sklearn.model_selection import cross_val_score
from sklearn.impute import SimpleImputer # missing values
from sklearn.preprocessing import StandardScaler # Scaling the data to have a mean of 0 and a standard deviation of 1 
from sklearn.pipeline import make_pipeline # Composing multiple steps behind each other
from sklearn.compose import make_column_transformer # Applying a transformer to a subset of columns
from sklearn.preprocessing import OneHotEncoder # Turning categorical data to numeric by making a column for each unique value and denoting it as 0 or 1
from sklearn.model_selection import train_test_split
```

Let's select some of the numerical and categorical features that we want to include in our model. We do so based on previous observations: LeadTime, TotalOfSpecialRequests and RequiredCarParkingSpaces had a relatively high correlation with IsCancelled, so it makes sense to include those. 
Furthermore, we might include some features because we think they might be relevant, even though evidence is not directly suggesting this, such as 'Babies'. People with babies might cancel more, for example because their baby is sick. 
In a later stage, we can leave out some of the features and see if our model is still performing well.

```python
num_features = ["LeadTime","ArrivalDateWeekNumber","ArrivalDateDayOfMonth",
                "StaysInWeekendNights","StaysInWeekNights","Adults","Children",
                "Babies","PreviousCancellations",
                "PreviousBookingsNotCanceled", "ADR"]

cat_features = ["hotel","ArrivalDateMonth","Meal","MarketSegment", "IsRepeatedGuest", "RequiredCarParkingSpaces", "TotalOfSpecialRequests",
                "DistributionChannel","ReservedRoomType","DepositType","CustomerType"]
```

```python
# Divide the data into X and y

X = hotel_df.drop(["IsCanceled"], axis=1)
y = hotel_df["IsCanceled"]
```

```python
preprocessor = make_column_transformer(
    (StandardScaler(), num_features),
    (OneHotEncoder(sparse_output=False, handle_unknown='ignore'), cat_features),
    remainder="drop"
)
```

Notice that we added "handle_unknown='ignore'" to the one hot encoder. This will ignore null values in our data, which we have for 'company', for example. If we do not add this statement, there will be errors in the code below.

```python
# Split the data into training and testing
X_train, X_test, y_train, y_test = train_test_split(X, y)
```

#### ❓Below, we have added some code that allows you to study the correlation of the variables after preprocessing. Take a look, what changed?

```python
# 1. Fit the preprocessor on your training data
preprocessor.fit(X_train)

# 2. Transform the data
X_processed = preprocessor.transform(X_train)

# 3. Convert to a DataFrame
#    If your preprocessor includes ColumnTransformer with named columns, you can get feature names
feature_names = preprocessor.get_feature_names_out()
X_processed_df = pd.DataFrame(X_processed, columns=feature_names)
# Add IsCancelled back
X_processed_df["IsCanceled"] = y_train.values

# 4. Compute the correlation matrix
corr_matrix = X_processed_df.corr()

# 5. (Optional) visualize or inspect correlations
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, cmap="coolwarm", center=0)
plt.title("Correlation Matrix after Preprocessing")
plt.show()
```

```python
abs(X_processed_df.corr(numeric_only=True)['IsCanceled']).sort_values(ascending=False)
```

We see that type of deposit, which was not in our correlation matrix before because it was categorical, now has the strongest correlation with cancellations. We will definetly include this variable in our final model. It would be a good idea for us to gradually reduce the number of features, dropping the ones with a lower correlation first, and comparing their performance.

## Modeling

#### ❓Take a look at the lecture slides. Which classification models have we seen in class? In a next step, we'll try to fit and evaluate all of those. Make sure to import them here.

```python
from sklearn.linear_model import LogisticRegressionCV # logistic regression for classification
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import HistGradientBoostingClassifier # Fits many decision trees sequentially. Each tree is trained to improve on the performance of all the previous trees.
from sklearn.ensemble import RandomForestClassifier # Fits many decision trees in parallel with some added randomness. The final prediction is the average of all trees..
```

In a next step, we'll fit different classification models and use cross validation to improve our models. We'll look at the average accuracy of these models.

### ❗ The parameter n_jobs specifies how many cross validations can be ran in parallel. Make sure to use n_jobs = -1 when doing cross validation.  This will allow you to run as many jobs simultanously as possible. Training these models can take a while.

```python
model_name_pair = [("logistic_regression", LogisticRegressionCV()), ("random_forest", RandomForestClassifier()), ("gradient boosting", HistGradientBoostingClassifier()), ("decision tree", DecisionTreeClassifier()) ]
pipes = []
for pair in model_name_pair:
    name, model = pair
    print(f"STARTING {name}")
    pipe = make_pipeline(preprocessor, model)
    pipes.append(pipe)
    results_acc = cross_val_score(pipe, X_train, y_train, cv=4, scoring="accuracy", n_jobs=-1) 
    results_roc = cross_val_score(pipe, X_train, y_train, cv=4, scoring="roc_auc", n_jobs=-1) 
    print("-"*20)
    print(f"The accuracy on the training set is {np.mean(results_acc)}\n")
    print(f"The auc on the training set is {np.mean(results_roc)} \n")
```

Repeat this process, but now with roc_auc as scoring metric. 

## Evaluating

### ❓ Motivate which model is the best and use it on the final part of the data


In this run random forest is the best performing model.

```python
from sklearn.metrics import accuracy_score
```

```python
# Fit the final model
rf_pipe = pipes[1]
rf_pipe.fit(X_train, y_train)
```

```python
# Make your predictions
y_pred = rf_pipe.predict(X_test)
y_pred
```

```python
# Calculate the final accuracy

accuracy_score(y_test, y_pred)
```

Finally, let's add a visualisation of the confusion matrix and roc curve:

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap=plt.cm.Blues)
plt.title("Confusion Matrix")
plt.show()
```

```python
from sklearn.metrics import roc_curve, auc

# 1. Get predicted probabilities for the positive class
y_prob = rf_pipe.predict_proba(X_test)[:, 1]

# 2. Compute FPR, TPR and thresholds
fpr, tpr, thresholds = roc_curve(y_test, y_prob)

# 3. Compute AUC
roc_auc = auc(fpr, tpr)

# 4. Plot ROC curve
plt.figure(figsize=(8,6))
plt.plot(fpr, tpr, color='blue', lw=2, label=f'ROC curve (AUC = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='red', lw=2, linestyle='--', label='Random Classifier')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.legend(loc='lower right')
plt.grid(alpha=0.3)
plt.show()
```
